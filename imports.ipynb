```python
#==================================
#============= IMPORTS ============
#==================================
from math import ceil, floor

# now I can call them without module.identifier (math.ceil)
print(ceil(123.44))
print(floor(454.645))
```

```python
# wild card imports are dangerous
# they import all identifiers from a module
# which could lead to unexpected errors

e = 'howdy'
print(e)

from math import *
print(e) # after import, this constant exists, replacing anything with the same name
```

```python
# binding names for modules and identifiers
import statistics as stats
print(stats.mean([11,245,23,254,22]))

from math import sqrt as sss # bind "sqrt" to "sss" instead
print(sss(144))
```
